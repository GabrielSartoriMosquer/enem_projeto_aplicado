## 📌 Notas Importantes

1. **NU_INSCRICAO**: Máscara do número original - não identifica a mesma pessoa em anos diferentes
2. **TP_FAIXA_ETARIA**: Calculada a partir da idade em 31/12/2025
3. **TRI**: Teoria de Resposta ao Item - escala normalizada entre edições (comparáveis entre anos)
4. **Questionário Socioeconômico**: Preenchimento obrigatório, mas alguns itens podem ter falta de resposta
5. **Todas as UFs**: Estão representadas com código IBGE de 2 dígitos (11-53)

## Utilização das bases de dados

#### PARTICIPANTES:

Métricas de participação (inscritos, taxas, cobertura geográfica)
Métricas geográficas (UF, município, região, urbano/rural)
Métricas demográficas (sexo, idade, raça, estado civil)
Métricas socioeconômicas (renda, escolaridade dos pais, bens)
Métricas educacionais (tipo de ensino, conclusão, tipo de escola)
Métricas de performance (notas por disciplina, distribuição)

#### ITENS_PROVA:

Métricas de escola (ranking, variância)
Métricas geográficas de escola (tipo de dependência)
Comparações institucionais (pública vs. privada)
Performance por disciplina
